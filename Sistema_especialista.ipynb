```python
hobbies = [
    {"Nome": "Ciclismo", "Tipo": "Esportes", "Ao Ar Livre": "Sim"},
    {"Nome": "Academia", "Tipo": "Esportes", "Ao Ar Livre": "Não"},
    {"Nome": "Pintura", "Tipo": "Arte", "Visual": "Visual"},
    {"Nome": "Violão", "Tipo": "Arte", "Visual": "Musical"},
]

regras = [
    {"Condição": {"Tipo": "Esportes", "Ao Ar Livre": "Sim"}, "Recomendação": ["Ciclismo"]},
    {"Condição": {"Tipo": "Esportes", "Ao Ar Livre": "Não"}, "Recomendação": ["Academia"]},
    {"Condição": {"Tipo": "Arte", "Visual": "Visual"}, "Recomendação": ["Pintura"]},
    {"Condição": {"Tipo": "Arte", "Visual": "Musical"}, "Recomendação": ["Violão"]},
]
```

```python
def recomendar_hobbies(preferencias):
    recomendacoes = []
    for regra in regras:
        condicao = regra["Condição"]
        if ("Tipo" in preferencias and "Tipo" in condicao and preferencias["Tipo"] == condicao["Tipo"]):
            if ("Ao Ar Livre" in preferencias and "Ao Ar Livre" in condicao and preferencias["Ao Ar Livre"] == condicao["Ao Ar Livre"]):
                recomendacoes.extend(regra["Recomendação"])
            elif ("Visual" in preferencias and "Visual" in condicao and preferencias["Visual"] == condicao["Visual"]):
                recomendacoes.extend(regra["Recomendação"])
    return recomendacoes
```

```python
def main():

    tipo = input("Escolha: Esportes ou Arte? ")

    if tipo == "Esportes":
        detalhe = input("Gosta de atividades ao ar livre? (Sim/Não) ")
        preferencias = {"Tipo": tipo, "Ao Ar Livre": detalhe}
    elif tipo == "Arte":
        detalhe = input("Escolha: Visual ou Musical? ")
        preferencias = {"Tipo": tipo, "Visual": detalhe}
    else:
        print("Opção inválida.")
        return


    hobbies_recomendados = recomendar_hobbies(preferencias)


    if hobbies_recomendados:
        print("Atividades de Lazer Recomendadas:")
        for hobby in hobbies_recomendados:
            print(f"- {hobby}")
    else:
        print("Nenhuma recomendação disponível.")

if __name__ == "__main__":
    main()
```
